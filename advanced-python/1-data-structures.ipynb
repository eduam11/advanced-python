## 1. Lists, Sets, Dictionaries + Comprehensions

### Lists

```python
mylist = [1, 2, 3]
```

```python
mylist.append() # Adds an element to the end of the list.
mylist.extend() # Extends the list by appending elements from an iterable.
mylist.insert() # Inserts an element at a specified position.
mylist.remove() # Removes the first occurrence of a value.
mylist.pop()	 # Removes and returns the element at the specified position.
mylist.clear() # Removes all elements from the list.
mylist.index() # Returns the index of the first occurrence of a value.
mylist.count() # Returns the number of occurrences of a value.
mylist.sort()	# Sorts the list in place.
mylist.reverse()	# Reverses the list in place.
mylist.copy()	# Returns a shallow copy of the list.
```

```python
mylist = [element for thing in things [if x]] # Comprehension
```

### Sets

```python
set = {1, 2, 2, 3, 3, 4} # Only unique elements
empty = set()
```

```python
set.issubset() # <= // Proper: <
set.issuperset() # >= // Proper: >
set.isdisjoint() # Check if no elements in common
set.add() # Only one element at a time
set.discard() # Remove one element
set.pop() # Extract and remove a random element
set.clear()
set.copy()
sorted(set) # Returns a LIST with the ordered set
```

```python
set = {element for thing in things [if x]} # Comprehension
```

### Dictionaries

```python
dict = {"key1": "value1", "key2": "value2"}
empty = {}
dict = dict()
dict = dict(**kwargs)
dict = dict(mapping, **kwargs)
dict = dict(iterable, **kwargs)
```

```python
dict.get(key, default) # Returns the value for a key, or a default if the key is not found.
dict.keys() # Returns a view of the dictionary's keys.
dict.values() # Returns a view of the dictionary's values.
dict.items() # Returns a view of the dictionary's key-value pairs.
dict.update() # Updates the dictionary with key-value pairs from another dict/iterable.
dict.pop(key, default) # Removes and returns the value for a specified key.
dict.popitem() # Removes and returns the last inserted key-value pair as a tuple (LIFO).
dict.setdefault() # Returns the value for a key, inserting it with a default if not present.
dict.clear() # Removes all items from the dictionary.
dict.copy() # Returns a shallow copy of the dictionary.
dict.fromkeys(iterable, value) # Creates a new dictionary from an iterable of keys, all set to a given value.
```

```python
dict = {key: value for thing in things [if x]} # Comprehension
```
